```python
from pydantic import BaseModel, EmailStr, Field
from typing import Optional
```

```python
class Student(BaseModel):

    name:str

names = {"name":"shashi"}

student = Student(**names)

print(student)
```

##### Default Values

```python
class Student(BaseModel):
    name: str = "shashi"

new_student = {}

student =  Student(**new_student)
print(student)
```

#### Optional

```python
class Student(BaseModel):
    name: str = "shashi"
    age: Optional[int] = None

new_student = {"age":39}

student = Student(**new_student)
print(student.name)
print(student.age)
    
```

#### Coerce

```python
class Student(BaseModel):
    name: str = "shashi"
    age: Optional[int] = None

new_student = {"age":"39"} # still it gets converted into int by pydantic 

student = Student(**new_student)
print(student.name)
print(student.age)
```

#### Built in Options

```python
class Student(BaseModel):
    name: str = "shashi"
    age: Optional[int] = None
    email :EmailStr

new_student = {"age":"39","email":"shahi@gmail.com"} # still it gets converted into int by pydantic 

student = Student(**new_student)
print(student.name)
print(student.age)
print(student.email)
```

#### Field Function

```python
class Student(BaseModel):
    name: str = "shashi"
    age: Optional[int] = None
    email :EmailStr
    cgpa : float = Field(gt=0, lt=10)

new_student = {"age":"39","email":"shahi@gmail.com", "cgpa":9}

student = Student(**new_student)
print(student)
```

```python
class Student(BaseModel):
    name: str = "shashi"
    age: Optional[int] = None
    email :EmailStr
    cgpa : float = Field(gt=0, lt=10, default=5)

new_student = {"age":"39","email":"shahi@gmail.com"}

student = Student(**new_student)
print(student)
```

#### Custome Description 

```python
class Student(BaseModel):
    name: str = "shashi"
    age: Optional[int] = None
    email :EmailStr
    cgpa : float = Field(gt=0, lt=10, default=5, description="A decimal value representing the cgpa of the student")

new_student = {"age":"39","email":"shahi@gmail.com", "cgpa":9}

student = Student(**new_student)


student_dict = dict(student)

student_dict["age"]
```
